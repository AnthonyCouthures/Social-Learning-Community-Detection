# Reproducing the figures of the paper

*Social learning community detection with nonlinear interaction*. A. Couthures, A. V. Jayakumar, V. S. Varma,
I.-C. Morărescu, S. Lasaulce and A. Girard.

This notebook goes through every figure of the paper. Each section runs the matching script of this folder
(`figXX_*.py`), which shows the figure here and saves it as a PDF in `paper/figures/`. The scripts can also be run
from a terminal, all at once with `python paper/make_all.py`.

**Setup.** From the repository root, run `pip install -e .` and `pip install -r paper/requirements.txt`, then open
this notebook from the `paper/` folder.

**Running time.** Most figures take seconds. The three parameter sweeps (Figs. 3, 4 and 9) take between a few
minutes and about 20 minutes on a laptop. By default this notebook re-plots them from the results stored in
`paper/results/`; set `RECOMPUTE = True` below to run them again.

```python
RECOMPUTE = False  # re-run the long sweeps of Figs. 3, 4 and 9 instead of using paper/results/
QUICK = False  # small, fast versions of every experiment (smoke test)

flags = "--quick" if QUICK else ""
sweep_flags = flags if RECOMPUTE else "--plot-only"
%matplotlib inline
```

## Fig. 1: Signal functions

Two signal functions with an unstable origin: the sigmoid $s(x) = \tanh(Kx)$ with $K = 2.5$, and the saturated
linear function $s(x) = \min(1, \max(-1, Kx))$ with $K = 4$. Their intersections with $y = x$ are the fixed points.

```python
%run fig01_signal_functions.py {flags}
```

## Fig. 2: SER on Zachary's karate club

(a) The network. (b) Edge reliability $P_{ij}$ with the symmetric *action vs. action* tension $|s_i - s_j|$.
(c) Directed reliability $P_{i\to j}$ with the *opinion vs. action* tension $|x_i - s_j|$.

```python
%run fig02_karate_ser.py {flags}
```

## Fig. 3: Splitting thresholds of LFR and ABCD benchmarks

The critical gain needed to split the whole graph, compared with the gains needed to split its ground-truth
communities, as the mixing parameter grows. A community structure can only be recovered while the global
threshold stays below the community thresholds. For ABCD graphs this holds up to $\xi \approx 0.5$.

```python
%run fig03_spectral_thresholds.py {sweep_flags}
```

conductance.
